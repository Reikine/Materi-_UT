```r
data <- read.csv('/home/indra/Dev/Personal/Materi-_UT/SMT_004_UT/SMT_004_Analisis_dan_Visualisasi_Data/tabel_tugas.csv')
data_bersih <- data[, c(2, 3)]
names(data_bersih) <- c('kecepatan', 'jarak')
kecepatan <- as.numeric(data_bersih$kecepatan)
jarak <- as.numeric(data_bersih$jarak)
```

### Jawaban Nomor 2

```r
# a
scatter_plot <- plot(
    kecepatan,
    jarak,
    main='Scatter Plot Jarak dan Kecepatan',
    col.main='red',
    col.lab='red',
    col.axis='red',
    fg='red',
    xlab='Kecepatan(km/jam)',
    ylab="Jarak(km)",
    pch=19,
    col='red'
)

scatter_plot
```

```r
# c
histogram <- hist(
    kecepatan,
    main='Histogram Kecepatan Mobil',
    xlab='Kecepatan(km/jam)',
    ylab='Frekuensi',
    col='red',
    border='red',
    col.main='red',
    col.lab='red',
    col.axis='red',
    fg='red',
)

histogram
```
